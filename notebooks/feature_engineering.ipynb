# Feature Engineering

This notebook prepares the state representation used by the reinforcement learning environment.

For every portfolio:

1. Technical indicators are calculated for each stock.
2. A rolling covariance matrix is computed using a 252-day lookback window.
3. Daily close prices are collected.
4. All features are combined into a single dataset.
5. The resulting dataset is saved for the reinforcement learning environment.

The generated dataset contains the complete state representation required by the PPO agent.

```python
from preprocessing import DataPreprocessor
from portfolio_builder import PortfolioBuilder
from feature_engineering import FeatureEngineer
```

```python
preprocessor = DataPreprocessor()

processed_data, train_data, _ = (
    preprocessor.load_processed_data()
)

builder = PortfolioBuilder()

portfolio_df = builder.load_portfolios()
```

```python
engineer = FeatureEngineer()

train_data = engineer.add_technical_indicators(
    train_data
)
```

```python
environment_data = engineer.build_state_dataset(
    train_data,
    portfolio_df,
)

environment_data.head()
```

```python
engineer.save_dataset(
    environment_data
)
```
