# Portfolio Construction

In this notebook, stocks are classified into three risk-based portfolios using the forecasted volatility estimated by the GARCH(1,1) model.

The procedure consists of:

1. Loading the estimated volatilities.
2. Sorting stocks by forecast volatility.
3. Dividing the stocks into three equally sized portfolios:
   - Conservative
   - Moderate
   - Aggressive
4. Saving the portfolio assignments for the reinforcement learning stage.

This classification follows the methodology described in the reference paper, where volatility serves as the primary measure of investment risk.

```python
from src.garch import GARCHModel
from src.portfolio_builder import PortfolioBuilder

garch = GARCHModel()

volatility_df = garch.load_results()

builder = PortfolioBuilder()
```

```python
portfolio_df = builder.build_portfolios(
    volatility_df
)

portfolio_df
```
