# Data Preprocessing

The objective of this notebook is to transform the raw historical market data into a clean and consistent dataset that can be used throughout the remaining stages of the project.

The preprocessing pipeline performs the following tasks:

- Remove duplicated records
- Handle missing values
- Sort observations by ticker and date
- Calculate daily simple returns
- Calculate daily logarithmic returns
- Split the dataset into training and testing periods
- Save the processed datasets for future use

The resulting datasets will be used in the volatility estimation, portfolio construction and reinforcement learning stages.

```python
from src.data_loader import DataLoader
from src.preprocessing import DataPreprocessor

loader = DataLoader()

raw_data = loader.get_data()

preprocessor = DataPreprocessor()

processed_data, train_data, test_data = preprocessor.preprocess(raw_data)
```
