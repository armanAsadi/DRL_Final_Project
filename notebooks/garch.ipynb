# GARCH(1,1) Volatility Estimation

This notebook estimates the conditional volatility of every stock in the training dataset using a GARCH(1,1) model.

For each stock:

1. A GARCH(1,1) model is fitted.
2. The model parameters are extracted.
3. The one-step-ahead conditional volatility is forecasted.
4. The results are stored in a single table for later use in portfolio construction.

The generated table will be used in the next stage of the project to classify stocks into different risk groups.

```python
from src.preprocessing import DataPreprocessor
from src.garch import GARCHModel

preprocessor = DataPreprocessor()

processed_data, train_data, test_data = preprocessor.load_processed_data()
```

```python
garch = GARCHModel()

volatility_df = garch.estimate_all(train_data)

volatility_df.head()
```

```python
garch.save_results(volatility_df)
```
