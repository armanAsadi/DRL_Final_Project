# Feature Scaling

This notebook standardizes the engineered features before training the reinforcement learning agent.

The objectives of this notebook are:

1. Load the processed train dataset.
2. Fit the feature scalers using training dataset.
3. Transform all training dataset using the fitted scalers.
4. Save the fitted scalers for future inference.
5. Save the scaled training dataset for the environment.

The scaler is fitted only on the training dataset to prevent data leakage. The learned statistics are then applied unchanged to the validation and test datasets.

```python
import pandas as pd

from feature_scaler import FeatureScaler
from feature_engineering import FeatureEngineer
```

```python
engineer = FeatureEngineer()

environment_data = engineer.load_dataset()
```

```python
scaler = FeatureScaler()

data = scaler.fit_transform(environment_data)
```

```python
scaler.save(
    "../models/feature_scalers.pkl"
)
```

```python
data.to_pickle(
    "../data/processed/environment_data_scaled.pkl"
)
```
